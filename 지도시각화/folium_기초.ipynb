```python
import pandas as pd
```

```python
location = "https://data.smartdublin.ie/dataset/33ec9fe2-4957-4e9a-ab55-c5e917c7a9ab/resource/2dec86ed-76ed-47a3-ae28-646db5c5b965/download/dublin.csv"
#자전거 대여소
bike_station_locations = pd.read_csv(location)
```

```python
bike_station_locations
```

```python
bike_station_locations = bike_station_locations[["Latitude", "Longitude", "Name"]] #위도, 경도, 이름
bike_station_locations
```

```python
map = folium.Map(location=[bike_station_locations.Latitude.mean(), bike_station_locations.Longitude.mean()], zoom_start=14, control_scale=True)
#위도와 경도의 평균 을 보여주게 지도를 보여주라. 예를 들어 경기도 유치원을 보여주고 싶으니까 경기도가 다 나오게 보여주라.
```

```python
map
```

```python
for index, location_info in bike_station_locations.iterrows(): #iterrows는 하나하나 내려가면서 돌려줘.
    folium.Marker([location_info["Latitude"], location_info["Longitude"]], popup=location_info["Name"]).add_to(map)
```

```python
map
```

```python
m = folium.Map(location=[37.607363, 127.042933], zoom_start=15, tiles="Stamen Terrain") #tiles 가 지도 종류이다.

folium.Marker(
    location=[37.607363, 127.042933],
    popup="Mt. Hood Meadows",
    icon=folium.Icon(icon="cloud"), #표시 마커도 여러 종류가 있다.
).add_to(m)

folium.Marker(
    location=[37.5999, 127.031111],
    popup="Timberline Lodge",
    icon=folium.Icon(color="green"),
).add_to(m)

folium.Marker(
    location=[37.6000, 127.02301],
    popup="Some Other Location",
    icon=folium.Icon(color="red", icon="info-sign"),
).add_to(m)


m
```

```python
m = folium.Map(location=[37.607363, 127.042933], tiles="Stamen Toner", zoom_start=13)

folium.Circle(
    radius=100,
    location=[37.607363, 127.042933],
    popup="The Waterfront",
    color="crimson",
    fill=False,
).add_to(m)

folium.CircleMarker(
    location=[37.607363, 127.042933],
    radius=50,
    popup="Laurelhurst Park",
    color="#3186cc",
    fill=True, # 원으로 채워준다.
    fill_color="#3186cc",
).add_to(m)


m
```

```python
m = folium.Map(location=[37.607363, 127.042933], tiles="Stamen Terrain", zoom_start=13)

m.add_child(folium.LatLngPopup())

m
```

# folium.PolyLine
## 폴리라인: 지하철 노선같은 라인을 표시하고 싶을때
## 폴리건: 구나 서울특별시 같은 큰 구역 표시하고 싶을때

```python
m = folium.Map(location=[-71.38, -73.9], zoom_start=11)   #이게 지하철 노선을 따라 가는 상권분석

trail_coordinates = [
    (-71.351871840295871, -73.655963711222626),
    (-71.374144382613707, -73.719861619751498),
    (-71.391042575973145, -73.784922248007007),
    (-71.400964450973134, -73.851042243124397),
    (-71.402411391077322, -74.050048183880477),
]


folium.PolyLine(trail_coordinates, tooltip="Coast").add_to(m)  #폴리움에 폴리라인이 있다. 저 좌표들을 집어넣으면 점들을 이어주는 선이 나타난다.

m
```

```python
import json

import requests

url = (
    "https://raw.githubusercontent.com/python-visualization/folium/main/examples/data"
)
vis1 = json.loads(requests.get(f"{url}/vis1.json").text)
vis2 = json.loads(requests.get(f"{url}/vis2.json").text)
vis3 = json.loads(requests.get(f"{url}/vis3.json").text)
```

```python
m = folium.Map(location=[46.3014, -123.7390], zoom_start=7, tiles="Stamen Terrain")

folium.Marker(
    location=[47.3489, -124.708],
    popup=folium.Popup(max_width=450).add_child(
        folium.Vega(vis1, width=450, height=250)
    ),
).add_to(m)

folium.Marker(
    location=[44.639, -124.5339],
    popup=folium.Popup(max_width=450).add_child(
        folium.Vega(vis2, width=450, height=250)
    ),
).add_to(m)

folium.Marker(
    location=[46.216, -124.1280],
    popup=folium.Popup(max_width=450).add_child(
        folium.Vega(vis3, width=450, height=250)
    ),
).add_to(m)

m
```
