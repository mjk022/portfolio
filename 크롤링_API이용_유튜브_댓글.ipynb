```python
import pandas
from googleapiclient.discovery import build
import pandas as pd
```

```python
comments = list()
api_obj = build('youtube', 'v3', developerKey="발급받은 구글 api")
response = api_obj.commentThreads().list(part='snippet,replies',
                                         videoId='zGPNQ8Vzxx0',
                                         maxResults=100).execute()
```

```python
while response:
    for item in response['items']:
        comment = item['snippet']['topLevelComment']['snippet']
        comments.append([comment['textDisplay'], comment['authorDisplayName'], comment['publishedAt'], comment['likeCount']])

        if item['snippet']['totalReplyCount'] > 0:
            for reply_item in item['replies']['comments']:
                reply = reply_item['snippet']
                comments.append([reply['textDisplay'], reply['authorDisplayName'], reply['publishedAt'], reply['likeCount']])

    if 'nextPageToken' in response:
        response = api_obj.commentThreads().list(part='snippet,replies', videoId='zGPNQ8Vzxx0', pageToken=response['nextPageToken'], maxResults=100).execute()
    else:
        break
```

```python
print(comments)
```

```python
df = pandas.DataFrame(comments)
df.to_excel("D:\results.xlsx", header=['comment', 'author', 'date', 'num_likes'], index=None)
```
