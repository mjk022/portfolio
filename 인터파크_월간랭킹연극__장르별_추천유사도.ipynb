##TF

```python
from sklearn.feature_extraction.text import CountVectorizer

#df 장르에 존재하는 공백 형식이 갖추어지지 않아있어서 공백있는거 공백 제거하기
df['장르']=df['장르'].str.replace(' ', '').str.replace(',', ', ')

#그 다음 ,를 기준으로 장르 나눠주기
df['genres_literal']=df['장르'].str.split(',')
df['genres_literal'] = df['genres_literal'].apply(lambda x : (' ').join(x))

count_vect = CountVectorizer(min_df=0, ngram_range=(1,2)) #우리가 지금 CountVectorizer 쓸거야.
genre_mat = count_vect.fit_transform(df['genres_literal'])

print(genre_mat.shape)
```

```python
df=df[['Title', '장르','genres_literal', 'numUsers', 'avgRating', 'stdRating']]
df
```

```python
print(genre_mat)
```

###코사인 유사도(TF)

```python
from sklearn.metrics.pairwise import cosine_similarity

genre_sim = cosine_similarity(genre_mat, genre_mat)#똑같은 메트릭스를 두개 집어넣는다.
print(genre_sim.shape)
```

```python
genre_sim_sorted_ind = genre_sim.argsort()[:, ::-1] #내림차순으로 sorting해서 보여줘.
print(genre_sim_sorted_ind[:1]) #영화 번호들 나옴
```

```python
def find_sim_movie(df, sorted_ind, title_name, top_n=10):

    # 인자로 입력된 movies_df DataFrame에서 'title' 컬럼이 입력된 title_name 값인 DataFrame추출
    title_movie = df[df['Title'] == title_name]

    # title_named을 가진 DataFrame의 index 객체를 ndarray로 반환하고
    # sorted_ind 인자로 입력된 genre_sim_sorted_ind 객체에서 유사도 순으로 top_n 개의 index 추출
    title_index = title_movie.index.values
    similar_indexes = sorted_ind[title_index, :(top_n)]

    # 추출된 top_n index들 출력. top_n index는 2차원 데이터 임.
    #dataframe에서 index로 사용하기 위해서 1차원 array로 변경
    print(similar_indexes)
    similar_indexes = similar_indexes.reshape(-1)

    return df.iloc[similar_indexes]
```

```python
similar_movies = find_sim_movie(df, genre_sim_sorted_ind, '(리얼타임 코믹연극) 택시안에서 - 부산',10)
similar_movies[['Title', 'avgRating']]
```

### 추천방식 1
몇 프로 이상되는 공연중에서만 추천해줘.

```python
percentile = 0.6
m = df['numUsers'].quantile(percentile)
C = df['avgRating'].mean()

def weighted_avgRating(record):
    v = record['numUsers']
    R = record['avgRating']

    return ( (v/(v+m)) * R ) + ( (m/(m+v)) * C )

df['weighted_vote'] = df.apply(weighted_avgRating, axis=1)
```

```python
df[['Title','avgRating','weighted_vote','numUsers']].sort_values('weighted_vote',
                                                                          ascending=False)[:10]
```

### 추천방식 2-1(TF)


```python
def find_sim_movie(df, sorted_ind, title_name, top_n=10):
    title_movie = df[df['Title'] == title_name]
    title_index = title_movie.index.values

    # top_n의 2배에 해당하는 쟝르 유사성이 높은 index 추출
    similar_indexes = sorted_ind[title_index, :(top_n*2)]
    similar_indexes = similar_indexes.reshape(-1)
    # 기준 영화 index는 제외
    similar_indexes = similar_indexes[similar_indexes != title_index]

    # top_n의 2배에 해당하는 후보군에서 weighted_vote 높은 순으로 top_n 만큼 추출
    return df.iloc[similar_indexes].sort_values('weighted_vote', ascending=False)[:top_n]
```

```python
similar_movies = find_sim_movie(df, genre_sim_sorted_ind, '(리얼타임 코믹연극) 택시안에서 - 부산',10)
similar_movies[['Title', 'avgRating', 'weighted_vote']]
```
